```python
import os
import pandas as pd
import shutil
```

```python
original_dataset = "/content/drive/MyDrive/Lettuce_Disease/lettuce disease.v1i.multiclass"

new_dataset = "/content/drive/MyDrive/LettuceCare_Dataset_2000"
```

```python
import os

print(os.path.exists(original_dataset))
```

```python
classes = [
    "Bacterial",
    "Powdery_mildew_on_lettuce",
    "Wilt_and_leaf_blight_on_lettuce",
    "fungal",
    "healthy"
]

print(classes)
```

```python
def get_label(row):

    for cls in classes:
        if row[cls] == 1:
            return cls

    return None
```

```python
def crop_dataset(source_split, target_split, jumlah_per_kelas):

    csv_path = os.path.join(
        original_dataset,
        source_split,
        "_classes.csv"
    )

    image_path = os.path.join(
        original_dataset,
        source_split
    )


    df = pd.read_csv(csv_path)

    df["label"] = df.apply(get_label, axis=1)


    for cls in classes:

        class_data = df[df["label"] == cls]


        class_data = class_data.sample(
            frac=1,
            random_state=42
        )


        selected_data = class_data.head(
            jumlah_per_kelas
        )


        save_folder = os.path.join(
            new_dataset,
            target_split,
            cls
        )


        os.makedirs(
            save_folder,
            exist_ok=True
        )


        for _, row in selected_data.iterrows():

            source_file = os.path.join(
                image_path,
                row["filename"]
            )


            destination_file = os.path.join(
                save_folder,
                row["filename"]
            )


            shutil.copy(
                source_file,
                destination_file
            )


        print(
            target_split,
            cls,
            "=",
            len(selected_data)
        )
```

```python
crop_dataset(
    "train",
    "train",
    320
)
```

```python
crop_dataset(
    "valid",
    "val",
    60
)
```

```python
crop_dataset(
    "test",
    "test",
    20
)
```

```python
for split in ["train","val","test"]:

    print("\n", split)

    split_path = os.path.join(
        new_dataset,
        split
    )


    for cls in classes:

        folder = os.path.join(
            split_path,
            cls
        )


        jumlah = len(
            os.listdir(folder)
        )

        print(cls, ":", jumlah)
```

```python
from ultralytics import YOLO
import os

# Memuat model terbaik hasil training
model = YOLO("/content/runs/classify/train/weights/best.pt")

# Target file uji coba
image_path = "/content/drive/MyDrive/test_daun.jpg"

# Jika file tidak ditemukan, gunakan gambar random dari dataset test sebagai contoh
if not os.path.exists(image_path):
    print(f"Gambar '{image_path}' tidak ditemukan.")
    print("Mencari gambar sampel alternatif dari test dataset...")
    test_dir = "/content/drive/MyDrive/LettuceCare_Dataset_2000/test"

    # Ambil sampel dari salah satu folder kelas
    sample_found = False
    for root, dirs, files in os.walk(test_dir):
        image_files = [f for f in files if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
        if image_files:
            image_path = os.path.join(root, image_files[0])
            print(f"Menggunakan sampel gambar: {image_path}")
            sample_found = True
            break

    if not sample_found:
        image_path = None
        print("Tidak ditemukan gambar sampel di folder test dataset.")

# Jalankan prediksi jika file gambar tersedia
if image_path and os.path.exists(image_path):
    result = model.predict(image_path)

    # Menampilkan hasil prediksi
    print("\nDaftar Kelas:", result[0].names)
    probs = result[0].probs
    predicted_class_id = probs.top1
    predicted_class_name = result[0].names[predicted_class_id]
    confidence = float(probs.top1conf)

    print(f"Hasil Prediksi: {predicted_class_name} ({confidence*100:.2f}%)")
else:
    print("Silakan unggah gambar daun lettuce ke Google Drive Anda dengan nama 'test_daun.jpg'.")
```
